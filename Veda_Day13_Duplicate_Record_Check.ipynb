```python
#Duplicate Records
duplicates = df.duplicated().sum()
duplicates
```

```python
order_duplicates = df[df.duplicated(subset=['Order ID'], keep=False)]

print("Duplicate Order ID Records:", len(order_duplicates))
order_duplicates.head()
```

```python
key_duplicates = df[
    df.duplicated(subset=['Order ID', 'Product ID'], keep=False)
]

print("Duplicate Order ID + Product ID:", len(key_duplicates))
key_duplicates.head()
```

```python
key_duplicates.sort_values(
    by=['Order ID', 'Product ID']
)
```

```python
key_duplicates[
    ['Order ID', 'Product ID', 'Customer ID',
     'Sales', 'Quantity', 'Discount', 'Profit']
]
```

```python
exact_duplicates = df[df.duplicated(keep=False)]

print("Total rows involved in exact duplicates:", len(exact_duplicates))
exact_duplicates
```

```python
duplicate_key_groups = (
    df[df.duplicated(
        subset=['Order ID', 'Product ID'],
        keep=False
    )][['Order ID', 'Product ID']]
    .drop_duplicates()
)

print("Total Duplicate Key Groups:", len(duplicate_key_groups))
```

```python
duplicate_report = (
    df[df.duplicated(
        subset=['Order ID', 'Product ID'],
        keep=False
    )]
    .sort_values(['Order ID', 'Product ID'])
)

duplicate_report.to_csv(
    "Veda_Day13_Duplicate_Report.csv",
    index=False
)

print("Duplicate Report created successfully!")
```

```python
cleaned_df = df.copy()

cleaned_df.to_csv(
    "Veda_Day13_Cleaned_Copy.csv",
    index=False
)

print("Cleaned Copy created successfully!")
print("Original Rows:", len(df))
print("Cleaned Rows:", len(cleaned_df))
```
